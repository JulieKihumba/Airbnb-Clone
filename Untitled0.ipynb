```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score

# Load dataset
file_path = "heart_attack_risk_dataset.csv"  # Use the correct file path
df = pd.read_csv(file_path)

# Display basic info and first few rows
print(df.info())  # Check data types and missing values
print(df.head())  # Preview dataset
```

```python
# Convert categorical columns using Label Encoding
label_encoders = {}
categorical_columns = ['Gender', 'Physical_Activity_Level', 'Stress_Level', 'Chest_Pain_Type',
                        'Thalassemia', 'ECG_Results', 'Heart_Attack_Risk']

for col in categorical_columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le  # Store encoders for future reference

# Define features (X) and target (y)
X = df.drop(columns=['Heart_Attack_Risk'])  # Independent variables
y = df['Heart_Attack_Risk']  # Target variable

# Standardize numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Split dataset into training (80%) and testing (20%) sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

print("Data preprocessing completed successfully.")
```

```python
# Initialize and train the model
model = RandomForestClassifier(n_estimators=100, random_state=42)
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Evaluate the model
print("Model Accuracy:", accuracy_score(y_test, y_pred))
print("Classification Report:\n", classification_report(y_test, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))
```

```python
def predict_heart_attack_risk(new_data):
    new_data = np.array(new_data).reshape(1, -1)

    # Ensure new data has the same number of features as the training set
    expected_features = X_train.shape[1]
    if new_data.shape[1] != expected_features:
        raise ValueError(f"Expected {expected_features} features, but got {new_data.shape[1]}.")

    new_data_scaled = scaler.transform(new_data)  # Apply scaling
    prediction = model.predict(new_data_scaled)
    return label_encoders['Heart_Attack_Risk'].inverse_transform(prediction)[0]

# Print expected number of features
print("Expected number of features:", X_train.shape[1])

# Correct sample patient data (adjust as needed)
sample_patient = [80, 0, 1, 1, 2, 25.6, 1, 1, 270.0, 100, 80, 1, 2, 1, 0, 1, 1, 150, 3]  # Ensure 19 values

predicted_risk = predict_heart_attack_risk(sample_patient)
print("Predicted Heart Attack Risk:", predicted_risk)
```
